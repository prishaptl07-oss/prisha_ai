# Data Frames
A data frame is a two dimensional table. It contains rows and columns. Similar to am excel spreadsheet

```python
data = {
    "Name": ["Salma", "Babu", "Nand", "Irfan"],
    "Math": [78, 34, 29, 33],
    "English": [58, 74, 69, 93],
    "Science": [68, 54, 59, 91]
}
students = pd.DataFrame(data)
print(students)
```

```python
print(students.dtypes) # Checking data types
print(students.shape) # Checking the shape of the data frame
```

Selecting rows: (loc[] select rows using labels)

```python
print(students.loc[0]) #selecting one row
print(students.loc[[1, 1]]) #Selecting multiple rows
#iloc[] selects rows using position numbers
print(students.iloc[0]) #Selecting one row
print(students.iloc[[0, 1]]) #Selecting multiple rows
```

```python
#selecting specific rows and columns
print(students.loc[0, "Name"]) #Selecting one row and one column
print(students.loc[[0, 1], ["Name", "Math"]]) #Multiple rows and columns
```

```python
#Filtering data
#students with more than 60 in maths
highscores = students[students["Math"] > 60]
print(highscores)
print(students["Math"].mean()) #average score in maths
print(students["Math"].min()) #minimum score in maths
print(students["Math"].max()) #maximum score in maths
print(students["Math"].sum()) #total score in maths
```
