```python
import matplotlib.pyplot as plt
```

```python
import pandas as pd
```

```python
students = pd.DataFrame({
    "Student": ["John", "Mary", "Ali", "Grace",
                "James", "Faith", "Brian", "Lucy"],
    "Age": [20,21,22,20,23,21,22,20],
    "Gender": ["M", "F", "M", "F",
               "M", "F", "M", "F"],
    "Math": [78,85,90,66,74,88,82,91],
    "Science": [80,79,95,70,88,90,76,94],
    "English": [75,88,92,72,81,86,79,90]
})
```

```python
print(students)
```

```python
#A line chart shows trends over time or ordered data.
plt.plot(students["Student"], students["Math"])
plt.show()
```

```python
#Adding titles
plt.plot(students["Student"], students["Math"])
plt.title("Math Scores of Students")
plt.xlabel("Students")
plt.ylabel("Math Scores")
plt.show()
```

```python
plt.plot(students["Student"], students["Science"])
plt.title("Science Scores of Students")
plt.xlabel("Students")
plt.ylabel("Science Scores")
plt.grid(True)
plt.show()
```
